## 1. A vector is a list of numbers

In machine learning a vector usually describes one thing: a house as `[area, rooms, age]`, or a person's ratings of four movies as `[5, 4, 0, 1]`. Adding two vectors adds them entry by entry; multiplying by a number multiplies every entry.

```python
def add(u, v):
    return [a + b for a, b in zip(u, v)]

def scale(c, v):
    return [c * a for a in v]

print(add([1, 2, 3], [10, 20, 30]))
print(scale(2, [1, -1, 0.5]))
```

## 2. The dot product

Multiply matching entries and add the results: `[1, 2, 3] · [4, 5, 6] = 1·4 + 2·5 + 3·6 = 32`. A large positive dot product means the vectors point in a similar direction; zero means they are perpendicular.

```python
u = [1, 2, 3]
v = [4, 5, 6]
steps = [a * b for a, b in zip(u, v)]
print(steps, sum(steps))
```

## 3. Length and cosine similarity

The length (norm) of a vector is the square root of its dot product with itself: `|[3, 4]| = 5`.

Cosine similarity divides the dot product by both lengths, so only the direction matters: `1` means the same direction, `0` perpendicular, `-1` opposite. Recommendation systems and text search use it to find similar items.

```python
import math

print(math.sqrt(3 * 3 + 4 * 4))

a = [1, 2]
b = [2, 4]   # same direction, twice as long
dot_ab = a[0] * b[0] + a[1] * b[1]
print(dot_ab / (math.sqrt(a[0]**2 + a[1]**2) * math.sqrt(b[0]**2 + b[1]**2)))
```

## 4. A matrix times a vector

A matrix is a table of numbers; in Python, a list of rows. To multiply a matrix by a vector, take the dot product of each row with the vector. A matrix with 2 rows and 3 columns ("2 by 3") needs a vector with 3 entries and gives a vector with 2 entries.

```python
A = [[1, 2, 0],
     [0, -1, 3]]
x = [2, 1, 1]
print([sum(a * b for a, b in zip(row, x)) for row in A])
```

## 5. The same with NumPy

NumPy does all of this in one line and much faster on large data. It is already installed in Colab and Kaggle; on your own computer run `pip install numpy` once.

`@` is the dot product for vectors and matrix multiplication for matrices.

```python
import numpy as np

u = np.array([1, 2, 3])
v = np.array([4, 5, 6])
print(u + v, 2 * u)
print(u @ v, np.linalg.norm(np.array([3, 4])))

A = np.array([[1, 2, 0], [0, -1, 3]])
x = np.array([2, 1, 1])
print(A.shape, A @ x)
```
