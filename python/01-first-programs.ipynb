## 1. `print` shows something on the screen

Text goes in quotes. You can print several things at once by separating them with commas; Python puts a space between them.

```python
print("Hello, world!")
print("I am learning Python.")
print(3, "apples and", 2, "pears")
```

## 2. Numbers and arithmetic

| Sign | Meaning | Example | Result |
|---|---|---|---|
| `+` `-` `*` | add, subtract, multiply | `7 * 3` | `21` |
| `/` | divide (always gives a decimal) | `7 / 2` | `3.5` |
| `//` | divide and drop the remainder | `7 // 2` | `3` |
| `%` | the remainder | `7 % 2` | `1` |
| `**` | power | `2 ** 3` | `8` |

Multiplication and division happen before addition and subtraction, just like in math class. Use parentheses to change the order.

```python
print(7 + 3, 7 - 3, 7 * 3)
print(7 / 2, 7 // 2, 7 % 2, 2 ** 3)
print(2 + 3 * 4, (2 + 3) * 4)
```

## 3. Variables: a name for a value

`=` means "store the value on the right under the name on the left". Later you can use the name instead of the value, and you can change what it stores.

```python
price = 4
quantity = 3
total = price * quantity
print("Total:", total)

quantity = quantity + 1   # one more item
print("New total:", price * quantity)
```

## 4. Strings: text in quotes

`+` joins strings, `*` repeats them, `len()` counts the characters. An f-string (a string with `f` before the quote) puts values inside text with `{ }`.

```python
first = "Ada"
greeting = "Hello, " + first + "!"
print(greeting)
print("ha" * 3)
print(len(first))
print(f"{first} has {len(first)} letters")
```

## 5. `input()`: the program asks, the user types

When you run a program yourself, `input("How many tickets? ")` waits for you to type an answer. To keep this notebook running from top to bottom without stopping, we store a typed answer in a variable instead. Try the real `input()` line yourself: remove the `#` in front of it.

Important: `input()` always gives **text**, even if you type a number. Use `int()` (whole numbers) or `float()` (decimals) to turn it into a number.

```python
# answer = input("How many tickets? ")
answer = "4"            # what a user might type: always text
print(answer * 2)       # text repeated: 44
tickets = int(answer)   # now a number
print(tickets * 12)     # 4 tickets at 12 dollars: 48
```
